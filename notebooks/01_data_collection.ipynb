```python
import pandas as pd
import requests

# Top 12 repositories across major languages (1 Request per Repo)
repos = [
    # Python (2)
    "pandas-dev/pandas", "psf/requests",
    # JavaScript (2)
    "facebook/react", "vuejs/vue",
    # Java (2)
    "spring-projects/spring-framework", "elastic/elasticsearch",
    # C++ (2)
    "tensorflow/tensorflow", "protocolbuffers/protobuf",
    # Go (2)
    "golang/go", "kubernetes/kubernetes",
    # Extra 2 top repos (TypeScript & Rust)
    "microsoft/vscode", "rust-lang/rust"
]

all_data = []

for repo in repos:
    url_repo = f"https://api.github.com/repos/{repo}"
    res_repo = requests.get(url_repo)
    
    print(f"Fetching {repo} -> Status: {res_repo.status_code}")
    
    if res_repo.status_code == 200:
        data = res_repo.json()
        
        repo_data = {
            "Repository": data["name"],
            "Stars": data["stargazers_count"],
            "Forks": data["forks_count"],
            "Open Issues": data["open_issues_count"],
            "Primary Language": data["language"] if data["language"] else "Unknown",
            "Last Updated": data["updated_at"]
        }
        all_data.append(repo_data)

# Create DataFrame
df = pd.DataFrame(all_data)

print(f"\nTotal Repositories Fetched: {len(df)}")
display(df)

# Save to CSV file
if not df.empty:
    df.to_csv("github_intelligence_data.csv", index=False)
    print("✅ CSV file successfully saved!")
```
